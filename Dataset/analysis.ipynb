```python
import pandas as pd
```

```python
df = pd.read_csv('transactions.csv')

df = df.drop(df.columns[[1,4,5,8]], axis=1) #drop the Reference ID, Till No, Account No and MM System columns
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst='True') #convert the Date column to datetime
```

```python
df['Time'] = df['Date'].dt.time #extract the time from the Date column
df['Date'] = df['Date'].dt.date #extract the date from the Date column
df.head()
```

```python
df = df.rename(columns={'Sender Identifier': 'ID'})
df = df.rename(columns={'Fees (KSh)': 'Fees'})
df = df.rename(columns={'Amount (KSh)': 'Amount'})
df.head()
```

```python
df = df.astype({'ID': str, 'Name': str, 'Fees':float}) #convert the ID and Name columns to string and the Fees and Amount columns to float
df.dtypes
```

```python
df['Amount'] = df['Amount'].str.replace(',','')
df['Amount'] = df['Amount'].astype(float) #convert the Amount column to float
```

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst='True') #convert the Date column to datetime
df.dtypes
```

```python
df['Time'] = df['Time'].astype(str) #convert the Time column to string
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M:%S') #convert the Time column to time

df.dtypes
```
